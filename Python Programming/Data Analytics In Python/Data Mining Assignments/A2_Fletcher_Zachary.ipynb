```python
auctioned_cars["over_70k_miles"] = auctioned_cars["VehOdo"] > 70000
filter_over_70K_true = auctioned_cars[auctioned_cars["over_70k_miles"] == True]
sort_by_vehodo = filter_over_70K_true.sort_values("VehOdo")

sort_by_vehodo.head()
```

```python
auctioned_cars.boxplot(column = ["VehOdo"])
plt.title("Boxplot of VehOdo")
plt.show()
```

Above is a boxplot showing the median, min, max, and outliers of VehOdo

```python
auctioned_cars.Size.value_counts().plot(kind = "bar")
plt.title("Bar Chart of Count of Car Size")
plt.show()
```

Above is a bar plot showing the counts of car size. Medium is the most common size to auction, while sports car is the least.



```python
auctioned_cars.WarrantyCost.plot(kind = "hist", bins = 6)
plt.title("Histogram of Warrenty Cost")
plt.show()
```

Above is a histogram of warranty cost. The histogram is not normally distributed, but instead skews to the left, with cheaper warranties being most common.

```python

sb.pairplot(auctioned_cars.drop(columns = "over_70k_miles"))
plt.show()
```

```python
sb.pairplot(auctioned_cars.drop(columns = "over_70k_miles"), hue = "IsBadBuy")
plt.show()
```

Above is a pair plot hued by bad buys. Bad buys seems to have a higher odometer and a lower buy cost.
